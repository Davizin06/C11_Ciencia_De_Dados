```python
import pandas as pd
import numpy as np

df_paises = pd.read_csv('paises.csv', delimiter=';')
df_space = pd.read_csv('space.csv', delimiter=';')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

ds_iris = sns.load_dataset('iris')
ds_setosa = ds_iris[ds_iris['species'] == 'setosa']

# calculando correlação
corr_setosa = ds_setosa.select_dtypes(include=['float64', 'int64']).corr()

# grafico
sns.heatmap(corr_setosa, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlação - Espécie Setosa (Iris)')
plt.show()
```

```python
ds_titanic = sns.load_dataset('titanic')

# Histograma com KDE
sns.histplot(data=ds_titanic, x='age', hue='sex', kde=True)
plt.title('Distribuição de Idades por Sexo (Titanic)')
plt.show()
```

```python
# boxplot com segmentação
sns.boxplot(data=ds_titanic, x='class', y='age', hue='sex')
plt.title('Idade por Classe e Sexo (Titanic)')
plt.show()
```

```python
ds_mpg = sns.load_dataset('mpg')

# grafico Regplot
sns.regplot(data=ds_mpg, x='horsepower', y='mpg', line_kws={'color': 'red'})
plt.title('Potência vs Consumo (MPG)')
plt.show()
```

```python
cols_ex5 = ['GDP ($ per capita)', 'Literacy (%)', 'Infant mortality (per 1000 births)', 'Phones (per 1000)']
corr_paises = df_paises[cols_ex5].corr()

sns.heatmap(corr_paises, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Matriz de Correlação (Países)')
plt.show()
```

```python
# Filtragem das duas regiões
regions_to_include = ['LATIN AMER. & CARIB', 'WESTERN EUROPE']
df_filtered = df_paises[df_paises['Region'].str.strip().isin(regions_to_include)]

sns.boxplot(data=df_filtered, x='Region', y='GDP ($ per capita)')
plt.title('PIB per capita: América Latina vs Europa Ocidental')
plt.show()
```

```python
sns.regplot(data=df_paises, x='Literacy (%)', y='Infant mortality (per 1000 births)', line_kws={'color': 'orange'})
plt.title('Alfabetização vs Mortalidade Infantil')
plt.show()
```

```python
df_space.columns = df_space.columns.str.strip()

#filtrando apenas missoes com valor conhecido
df_space_cost = df_space[df_space['Cost'] > 0]

# gerando o grafico
sns.boxplot(data=df_space_cost, x='Status Rocket', y='Cost')
plt.title('Custo das Missões: Foguetes Ativos vs Aposentados')
plt.show()
```
