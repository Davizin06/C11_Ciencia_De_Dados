```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Carregando os datasets
df_paises = pd.read_csv('paises.csv', delimiter=';')
df_space = pd.read_csv('space.csv', delimiter=';')
```

PARTE 1

```python
# Filtrando os paises da america do norte
df_na = df_paises[df_paises['Region'].str.strip() == 'NORTHERN AMERICA']

# Plotando
plt.plot(df_na['Country'], df_na['Deathrate'], 'r-', df_na['Country'], df_na['Birthrate'], 'b--')
plt.xlabel('Países da América do Norte')
plt.ylabel('Taxas')
plt.title('Mortalidade (Vermelha) x Natalidade (Azul)')
plt.xticks(rotation=45)
plt.show()
```

```python
# Filtrando as missoes por localizacao
df_eua = df_space[df_space['Location'].str.contains('USA')]
df_china = df_space[df_space['Location'].str.contains('China')]

# Contando a quantidade de empresas
qt_eua = len(df_eua['Company Name'].unique())
qt_china = len(df_china['Company Name'].unique())

# grafico de barras
plt.bar(['EUA', 'China'], [qt_eua, qt_china], color='blue')
plt.title('Empresas Espaciais (EUA vs China)')
plt.show()
```

```python
# Extraindo dados apenas da Roscosmos
df_roscosmos = df_space[df_space['Company Name'] == 'Roscosmos']

qt_sucesso = len(df_roscosmos[df_roscosmos['Status Mission'] == 'Success'])
qt_falha = len(df_roscosmos[df_roscosmos['Status Mission'] == 'Failure'])

plt.pie(x=[qt_sucesso, qt_falha], labels=['Sucesso', 'Falha'], autopct='%1.1f%%')
plt.title('Missões da Roscosmos')
plt.show()
```

```python
df_falhas = df_space[df_space['Status Mission'] == 'Failure']
top_5_falhas = df_falhas['Company Name'].value_counts().head(5)

plt.bar(top_5_falhas.index, top_5_falhas.values, color='red')
plt.title('Top 5 Empresas com mais Falhas')
plt.xticks(rotation=45)
plt.show()
```

PARTE 2

```python
df_latam = df_paises[df_paises['Region'].str.strip() == 'LATIN AMER. & CARIB']

plt.scatter(df_latam['GDP ($ per capita)'], df_latam['Literacy (%)'], s=df_latam['Population']/100000)
plt.xlabel('GDP ($ per capita)')
plt.ylabel('Literacy (%)')
plt.title('Renda x Alfabetização (LATAM)')
plt.show()
```

```python
qt_ativos = len(df_space[df_space['Status Rocket'] == 'StatusActive'])
qt_aposentados = len(df_space[df_space['Status Rocket'] == 'StatusRetired'])

plt.pie(x=[qt_ativos, qt_aposentados], labels=['Ativos', 'Aposentados'], autopct='%1.1f%%')
plt.title('Status dos Foguetes')
plt.show()
```

```python
df_eu = df_paises[df_paises['Region'].str.strip() == 'WESTERN EUROPE']

plt.plot(df_eu['Country'], df_eu['GDP ($ per capita)'], 'o-b', df_eu['Country'], df_eu['Phones (per 1000)'], 's--r')
plt.xticks(rotation=90)
plt.title('PIB x Telefones (Europa Ocidental)')
plt.show()
```

```python
top_5_sucesso = df_space[df_space['Status Mission'] == 'Success']['Company Name'].value_counts().head(5)
top_5_falha = df_space[df_space['Status Mission'] == 'Failure']['Company Name'].value_counts().head(5)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('Top 5 Sucessos')
plt.bar(top_5_sucesso.index, top_5_sucesso.values, color='green')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
plt.title('Top 5 Falhas')
plt.bar(top_5_falha.index, top_5_falha.values, color='red')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
