```python
import pandas as pd

dfPaises = pd.read_csv('paises.csv', sep=';')

#quais são os países da OCEANIA
paises_oceania = dfPaises[dfPaises['Region'].str.contains('OCEANIA')]
print("Países da Oceania:\n", paises_oceania['Country'].values, "\n")

# quantos países são da OCEANIA
qtd_oceania = len(paises_oceania)
print("Quantidade de países na Oceania:", qtd_oceania, "\n")

#País com a maior população
indice_maior = dfPaises['Population'].idxmax()
pais_maior_populacao = dfPaises.loc[indice_maior, ['Country', 'Region']]
print("País com maior população:\n", pais_maior_populacao, "\n")

# agrupando por regiao e mostrando media de alfabetizacao
media_alfabetizacao = dfPaises.groupby('Region')['Literacy (%)'].mean()
print("Média de alfabetização por região:\n", media_alfabetizacao, "\n")

# paises que possuem costa maritima
paises_sem_costa = dfPaises[dfPaises['Coastline (coast/area ratio)'] == 0][['Country']]
paises_sem_costa.to_csv('noCoast.csv', index=False, sep=';')
print(f"Quantidade de países sem costa: {len(paises_sem_costa)}\n")

# funcao para mostrar taxa de mortalidade
def define_status_ajuda(taxa_mortalidade):
    return 'Balanced' if taxa_mortalidade < 9 else 'Urgent'

dfPaises['Humanitarian Help'] = dfPaises['Deathrate'].apply(define_status_ajuda)
print("Dataset com a nova coluna:")
print(dfPaises[['Country', 'Deathrate', 'Humanitarian Help']].head())
```

```python
#agrupando por região
estatisticas_populacao = dfPaises.groupby('Region')['Population'].describe()
print("Estatísticas da População (5 primeiras linhas):\n", estatisticas_populacao.head(5), "\n")

#funcao para reduzir a mortalidade infantil em 15%
def reduzindo_mortalidade(valor):
    return valor * 0.85

mortalidade_original = dfPaises['Infant mortality (per 1000 births)']
mortalidade_reduzida = mortalidade_original.apply(reduzindo_mortalidade)

comparacao = pd.concat([mortalidade_original, mortalidade_reduzida], axis=1)
comparacao.columns = ['Mortalidade Original', 'Mortalidade Reduzida (-15%)']
print("Comparação da Mortalidade Infantil (primeiros registos):\n", comparacao.head(), "\n")

# remover a coluna Coastline e guardar o dataset
dfPaises_sem_coastline = dfPaises.drop('Coastline (coast/area ratio)', axis=1)
dfPaises_sem_coastline.to_csv('paises_sem_coastline.csv', sep=';', index=False)
print("Dataset sem a coluna Coastline criado")
```
