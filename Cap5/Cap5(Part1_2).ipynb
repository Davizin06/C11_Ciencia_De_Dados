```python
import pandas as pd

series1 = pd.Series({'Java': 16.25, 'C': 16.04, 'Python': 9.85})
series2 = pd.Series({'C': 16.21, 'Python': 12.12, 'Java': 11.68})

total_ano1 = series1.sum()
total_ano2 = series2.sum()
print("Porcentagem total (Ano 1):", total_ano1)
print("Porcentagem total (Ano 2):", total_ano2, "\n")

crescimento = series2 - series1
print("Crescimento/Declínio:\n", crescimento, "\n")

linguagens_crescimento = crescimento[crescimento > 0]
print("Linguagens com crescimento:\n", linguagens_crescimento, "\n")

projecao = series2 + (2 * crescimento)
linguagem_mais_popular = projecao.nlargest(1)
print("Linguagem mais popular em 2 anos:\n", linguagem_mais_popular)
```

```python
import numpy as np

np.random.seed(10)
index = ['A', 'B', 'C', 'D', 'E']
columns = ['W', 'X', 'Y', 'Z']
df = pd.DataFrame(data=np.random.randint(1, 50, [5, 4]), index=index, columns=columns)
print("DataFrame Base:\n", df, "\n")

media_x_menor_30 = df['X'][df['X'] < 30].mean()
print("Média (Coluna X < 30):", media_x_menor_30, "\n")

media_d = df.loc['D'].mean()
soma_e = df.iloc[4].sum()
print("Média da linha D:", media_d)
print("Soma da linha E:", soma_e, "\n")

df_slice = df.loc[['A', 'C', 'E'], ['X', 'Y']]
print("Matriz (Slicing):\n", df_slice, "\n")

soma_linhas = df_slice.sum(axis=1)
soma_colunas = df_slice.sum(axis=0)
print("Soma por linhas:\n", soma_linhas, "\n")
print("Soma por colunas:\n", soma_colunas)
```
