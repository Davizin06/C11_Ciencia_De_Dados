Slicing no Numpy

```python
import numpy as np
```

```python
#Plantando a mesma semente aleatoria
np.random.seed(10)
# Estruturando uma matriz 3x3
mtz = np.random.randint(1, 99, 9).reshape(3, 3) #randomrandint para criar com numeros aleatorios e .reshape para transformar em matriz
print(mtz)
```

```python
# Extraindo apenas a segunda linha da matriz
print(mtz[1])

#Extraindo apenas a terceira coluna da matriz
print(mtz[:,2])

#Extraindo a matriz 2x2 no canto inferior direito da matriz
print(mtz[1:,0:2])
```

CONDICIONAIS NO NUMPY

```python
print(mtz)

#Mostre apenas os elementos menores que 70
# print(mtz<70)
# print(mtz[mtz<70])

#Retorne apenas os numeros pares
print(mtz%2==0)
print(mtz[mtz%2==0])
```

ANÁLISE DE PADRÕES TEXTUAIS COM NUMPY

```python
arr = np.array(['Inatel', 'Casa Viva', 'ICC',
               'CDG', 'CSILab', 'RobotBulls'])
print(arr)

cond = np.char.find(arr,'a') >=0
#Submodulo de NumPy para trabalhar com texto: char
#Buscando qual texto aceita um padrão informado

arr = np.char.upper(arr)
print(np.char.find(arr,'a'))
print(cond)
print(arr[cond])
```
