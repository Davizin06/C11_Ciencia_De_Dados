```python
import numpy as np

dataset = np.loadtxt('space.csv', delimiter=';', dtype=str, encoding='utf-8', skiprows=1)

COL_EMPRESA = 1
COL_LOCAL = 2
COL_CUSTO = 6
COL_STATUS_MISSAO = 7
COL_STATUS_FOGUETE = 5

print("EXERCÍCIO 1")
status_missoes = dataset[:, COL_STATUS_MISSAO]
missoes_sucesso = np.sum(status_missoes == 'Success')
porcentagem_sucesso = (missoes_sucesso / len(status_missoes)) * 100
print(f"Porcentagem de sucesso: {porcentagem_sucesso:.2f}%")

print("\nEXERCÍCIO 2")
custos_str = np.char.strip(np.char.replace(dataset[:, COL_CUSTO], ',', ''))
gastos_validos_str = custos_str[custos_str != '']
gastos_float = gastos_validos_str.astype(float)
gastos_maiores_que_zero = gastos_float[gastos_float > 0]

media_gastos = gastos_maiores_que_zero.mean()
print(f"Média de gastos das missões: {media_gastos:.2f}")

print("\nEXERCÍCIO 3")
locais = dataset[:, COL_LOCAL]
eua_condicao = np.char.find(locais, 'USA') != -1
qtd_eua = np.sum(eua_condicao)
print(f"Missões realizadas pelos EUA: {qtd_eua}")

print("\nEXERCÍCIO 4")
empresas = dataset[:, COL_EMPRESA]
condicao_spacex = empresas == 'SpaceX'
custos_spacex_str = custos_str[condicao_spacex]
custos_spacex_validos = custos_spacex_str[custos_spacex_str != ''].astype(float)

missao_mais_cara_spacex = custos_spacex_validos.max()
print(f"Missão mais cara da SpaceX: {missao_mais_cara_spacex}")

print("\nEXERCÍCIO 5")
empresas_unicas, contagem_missoes = np.unique(empresas, return_counts=True)
print("Empresas e quantidade de missões realizadas:")
for empresa, qtd in zip(empresas_unicas, contagem_missoes):
    print(f" - {empresa}: {qtd}")
```

```python
print("\nEXERCÍCIO 6")
status_foguete = dataset[:, COL_STATUS_FOGUETE]
foguetes_aposentados = np.sum(status_foguete == 'StatusRetired')
porcentagem_aposentados = (foguetes_aposentados / len(status_foguete)) * 100
print(f"Porcentagem de foguetes com status 'StatusRetired': {porcentagem_aposentados:.2f}%")

print("\nEXERCÍCIO 7")
condicao_russia = np.char.find(locais, 'Russia') != -1
qtd_russia = np.sum(condicao_russia)
print(f"Missões lançadas a partir da Rússia: {qtd_russia}")

print("\nEXERCÍCIO 8")

custos_zeros = np.where(custos_str == '', '0', custos_str)
custos_float_todas = custos_zeros.astype(float)

indice_mais_caro = custos_float_todas.argmax()
empresa_mais_cara = dataset[indice_mais_caro, COL_EMPRESA]
valor_mais_caro = custos_float_todas.max()

print(f"A missão mais cara foi da empresa {empresa_mais_cara}, custando: {valor_mais_caro}")
```
