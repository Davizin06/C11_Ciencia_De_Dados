```python
import pandas as pd
import numpy as np

# Carregando o dataset
df = pd.read_csv('paises.csv', delimiter=';')

# Limpeza inicial
df['Region'] = df['Region'].str.strip()
```

```python
print("QUESTÃO 1")

# Fazendo o slicing

df_q1 = df[['Country', 'Region', 'Population', 'Area (sq. mi.)']]
print(df_q1.head(10))
```

```python
print("QUESTÃO 2")

# Obtendo as regiões únicas
regioes_unicas = df['Region'].unique()
quantidade_regioes = len(regioes_unicas)

print(f"Existem {quantidade_regioes} regiões diferentes no dataset. São elas:")
for regiao in regioes_unicas:
    print(f"- {regiao}")
```

```python
print("QUESTÃO 3")

# Calculando a média da coluna
media_alfabetizacao = df['Literacy (%)'].mean()
print(f"A taxa média de alfabetização no planeta é de: {media_alfabetizacao:.2f}%")
```

```python
print("QUESTÃO 4")

# Filtrando o dataset
paises_america_norte = df[df['Region'] == 'NORTHERN AMERICA']
quantidade_america_norte = len(paises_america_norte)
print(f"Existem {quantidade_america_norte} países na América do Norte segundo o dataset.")
```

```python
print("QUESTÃO 5")

df_latam = df[df['Region'] == 'LATIN AMER. & CARIB']

indice_maior_gdp = df_latam['GDP ($ per capita)'].idxmax()

pais_maior_gdp = df.loc[indice_maior_gdp, 'Country']
valor_maior_gdp = df.loc[indice_maior_gdp, 'GDP ($ per capita)']

print(f"O país da América do Sul/Caribe com maior renda per capita é: {pais_maior_gdp.strip()}")
print(f"Valor do GDP ($ per capita): {valor_maior_gdp}")
```
