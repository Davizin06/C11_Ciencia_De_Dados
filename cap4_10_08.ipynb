**NumPy**

Criando um NumPy Array



```python
import numpy as np

##Criando um numpy array em cima de um vetor

arr = np.array([10,20,30,40,50,60])
print(arr)

#Propriedades do array
print(arr.size)
print(arr.ndim)
print(arr.shape)
print(type(arr))

##Criando np array de uma matriz 2 dimensoes

mtz = np.array([[ 10,20],[30,40],[50,60]])
print(mtz.size)
print(mtz.ndim)
print(mtz.shape)
print(type(arr))

```

**Funções pré-prontas para estruturar numpy arrays**

```python
#Ones
mtz = np.ones([5,5])
print(mtz)

#Zeros
arr = np.zeros([10])
print(arr.reshape([5,2])) ##Reshape para alterar a dimensao do array

#Arange
mtz = np.arange(2, 21, 2)

print(mtz.reshape(2,5))

```

**Operações entre numpy arrays**

```python
arr1 = np.array([10,20,30,40,50])
arr2 = np.array([60,70,80,90,40])

print(arr1 + arr2)


#Concatenacao de array

arr3 = np.concatenate([arr1,arr2])
print(arr3)

```

**Numeros aleatrios**

```python
##Módulo random

arr = np.random.randint(10)
print(arr)
```
