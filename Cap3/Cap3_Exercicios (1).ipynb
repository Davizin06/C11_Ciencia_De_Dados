```python
pessoas = []

for i in range(3):
    nome = input(f"Nome da pessoa {i+1}: ")
    peso = float(input(f"Peso da pessoa {i+1}: "))
    pessoas.append([nome, peso])

mais_pesada = pessoas[0]
mais_leve = pessoas[0]

for pessoa in pessoas:
    if pessoa[1] > mais_pesada[1]:
        mais_pesada = pessoa
    if pessoa[1] < mais_leve[1]:
        mais_leve = pessoa

print(f"Pessoa mais pesada: {mais_pesada[0]} ({mais_pesada[1]}kg)")
print(f"Pessoa mais leve: {mais_leve[0]} ({mais_leve[1]}kg)")
```

```python
n = int(input("Quantas pessoas deseja cadastrar? "))

soma_idades = 0
mulheres_sub20 = 0

for i in range(n):
    nome = input(f"Nome {i+1}: ")
    idade = int(input(f"Idade {i+1}: "))
    sexo = input(f"Sexo {i+1} (M/F): ").upper()

    soma_idades += idade

    if sexo == 'F' and idade < 20:
        mulheres_sub20 += 1

if n > 0:
    media_idade = soma_idades / n
else:
    media_idade = 0

print(f"Média de idade do grupo: {media_idade:.1f} anos")
print(f"Quantidade de mulheres com menos de 20 anos: {mulheres_sub20}")
```

```python
ingredientes = ['Farinha', 'Açúcar', 'Ovos', 'Leite']
print(f"Receita original: {ingredientes}")

#Adiciona no final
ingredientes.append('Fermento')

#Insere em posição específica
ingredientes.insert(1, 'Manteiga')

#Remove pelo valor
ingredientes.remove('Ovos')

print(f"Receita final: {ingredientes}")
```

```python
ingredientes_receita = {'Farinha', 'Manteiga', 'Açúcar', 'Leite', 'Fermento'}

pessoa1 = {'Farinha', 'Leite'}
pessoa2 = {'Manteiga', 'Açúcar', 'Ovos'}

falta_pessoa1 = ingredientes_receita - pessoa1
falta_pessoa2 = ingredientes_receita - pessoa2

print(f"Falta para a Pessoa 1: {falta_pessoa1}")
print(f"Falta para a Pessoa 2: {falta_pessoa2}")
```

```python
lista_produtos = []

for i in range(3):
    prod = {}
    prod['nome'] = input(f"Nome do produto {i+1}: ")
    prod['preco'] = float(input(f"Preço do produto {i+1}: "))
    prod['quantidade'] = int(input(f"Quantidade em estoque: "))

    lista_produtos.append(prod)

print("\nRelatório de Estoque")
for p in lista_produtos:
    valor_total = p['preco'] * p['quantidade']
    print(f"Produto: {p['nome']} | Valor total em estoque: R$ {valor_total:.2f}")
```
